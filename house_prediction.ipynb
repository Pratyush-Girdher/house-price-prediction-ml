```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv('train.csv')
test_data =pd.read_csv('test.csv')
```

```python
#defining the target(y) and features(X)
y=data.SalePrice
features=['MSSubClass','YearBuilt','YrSold','LotArea','OverallQual']
test_X=test_data[features]
X=data[features]
print(X)
```

```python
#splitintg the data
from sklearn.model_selection import train_test_split
train_X,val_X,train_y,val_y=train_test_split(X,y,random_state=0)
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error
def get_mae(max_leaf_nodes,train_X,val_X,train_y,val_y):
    data_model=DecisionTreeRegressor(max_leaf_nodes=max_leaf_nodes,random_state=0)
    data_model.fit(train_X,train_y)
    pred=data_model.predict(val_X)
    mae=mean_absolute_error(val_y,pred)
    return mae
```

```python
candidate_leaf_values=[5,50,100,250,500,1000,2500,5000]

score={leaf_value:get_mae(leaf_value,train_X,val_X,train_y,val_y) for leaf_value in candidate_leaf_values}
print(min(score,key=score.get))
```

```python
data_model=DecisionTreeRegressor(max_leaf_nodes=50,random_state=0)
data_model.fit(X,y)
predictions =data_model.predict(test_X)
```
